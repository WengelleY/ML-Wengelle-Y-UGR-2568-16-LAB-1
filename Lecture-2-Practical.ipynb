```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

n = 100

study_hours = np.random.normal(5, 2, n)
attendance = np.random.normal(85, 10, n)
previous_mark = np.random.normal(70, 15, n)

final_score = (
    (0.5 * study_hours)
    + (0.3 * attendance)
    + (0.2 * previous_mark)
    + np.random.normal(0, 5, n)
)

df = pd.DataFrame({
    'study_hours': study_hours,
    'attendance': attendance,
    'previous_mark': previous_mark,
    'final_score': final_score
})

df.loc[0:5, 'study_hours'] = np.nan
df.loc[99, 'study_hours'] = 50
df.loc[98, 'attendance'] = -10

print(df)


print("Outlier record:")
print(df[df['study_hours'] == 50])

print("\nMissing values:")
print(df.isnull().sum())

df['study_hours'] = df['study_hours'].fillna(
    df['study_hours'].median()
)

df['study_hours'] = df['study_hours'].clip(upper=12)

df.loc[df['attendance'] < 0, 'attendance'] = np.nan
df['attendance'] = df['attendance'].fillna(
    df['attendance'].median()
)

print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nCleaned dataset:")
print(df)



X = df.drop(columns=['final_score'])

y = df['final_score']

print("Feature matrix shape:", X.shape)
print("\nNumerical columns:")
print(X.select_dtypes(include=np.number).columns.tolist())

print("\nTarget variable:", y.name)
print("Target shape:", y.shape)




print(df.describe().loc[['mean', '50%', 'std']])

summary = df.describe().loc[['mean', '50%', 'std']]
summary = summary.rename(index={'50%': 'median'})

print("\nSummary statistics:")
print(summary.round(2))




plt.figure(figsize=(8, 5))

plt.hist(
    df['final_score'],
    bins=10,
    edgecolor='black'
)

plt.title('Distribution of Final Scores')
plt.xlabel('Final Score')
plt.ylabel('Number of Students')
plt.grid(axis='y', alpha=0.3)

plt.show()


plt.figure(figsize=(8, 5))

plt.scatter(
    df['study_hours'],
    df['final_score'],
    alpha=0.7
)

plt.title('Study Hours vs Final Score')
plt.xlabel('Study Hours')
plt.ylabel('Final Score')
plt.grid(alpha=0.3)

plt.show()




correlation = df['study_hours'].corr(df['final_score'])

print("Pearson correlation coefficient:", round(correlation, 4))




x = df.loc[0, ['study_hours', 'attendance', 'previous_mark']].to_numpy(dtype=float)

w = np.array([0.5, 0.2, 0.4])
b = 5

dot_product = np.dot(x, w)
prediction = dot_product + b

print("Feature vector x:", x)
print("Weight vector w:", w)
print("Bias b:", b)
print("Dot product:", round(dot_product, 2))
print("Predicted score:", round(prediction, 2))




for column in ['study_hours', 'previous_mark']:
    mean = df[column].mean()
    std = df[column].std()

    df[column + '_scaled'] = (df[column] - mean) / std

print(df[[
    'study_hours',
    'study_hours_scaled',
    'previous_mark',
    'previous_mark_scaled'
]].head())

print("\nStandardized means:")
print(df[['study_hours_scaled', 'previous_mark_scaled']].mean())

print("\nStandardized standard deviations:")
print(df[['study_hours_scaled', 'previous_mark_scaled']].std())

```
